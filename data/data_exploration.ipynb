# HYG.3: Vitamin & Supplement Intelligence - Exploratory Data Analysis

This notebook performs initial exploratory data analysis on the datasets requested for the MVP:
1. Vitamin Deficiency Disease Prediction Dataset
2. Pharmaceutical Drugs and Vitamins Dataset V2
3. Drugs and Vitamins Classification Dataset
4. Supplement Sales Data


```python
import pandas as pd
import kagglehub
from kagglehub import KaggleDatasetAdapter
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

# 1. Vitamin Deficiency Disease Prediction Dataset
path1 = kagglehub.dataset_download("nudratabbas/vitamin-deficiency-disease-prediction-dataset")
print("Vitamin Deficiency Dataset Path:", path1)

# 2. Pharmaceutical Drugs and Vitamins Dataset V2
path2 = kagglehub.dataset_download("vencerlanz09/pharmaceutical-drugs-and-vitamins-dataset-v2")
print("Pharmaceutical Drugs Dataset Path:", path2)

# 3. Drugs and Vitamins Classification
path3 = kagglehub.dataset_download("utkarshsaxenadn/drugs-and-vitamins-classification")
print("Drugs and Vitamins Classification Dataset Path:", path3)

# 4. Supplement Sales Data
path4 = kagglehub.dataset_download("zahidmughal2343/supplement-sales-data")
print("Supplement Sales Data Path:", path4)
```

## 1. Vitamin Deficiency Disease Prediction Dataset
Let's load the CSV file from the dataset and look at the first few rows and basic statistics.

```python
import os

# Load Vitamin Deficiency dataset
vitamin_csv_path = os.path.join(path1, "vitamin_deficiency_disease_dataset_20260123.csv")
df_vitamin = pd.read_csv(vitamin_csv_path)

display(df_vitamin.head())
display(df_vitamin.info())
```

## 2. Supplement Sales Data
Next, let's explore the Supplement Sales Data.

```python
# Load Supplement Sales Data
sales_csv_path = os.path.join(path4, "Supplement_Sales_Weekly_Expanded.csv")
df_sales = pd.read_csv(sales_csv_path)

display(df_sales.head())
display(df_sales.describe())
```
